#INDEXING AND SLICING


```python
print( ages[0])
print( names[0])
print( ages[-1])
print( names[-1])
print( names[1:4] ) #last value excluded
print( names[::1] )
ages[ 3 ] = 18 ;
print( ages[3] )
names.append("Vani")
print(names)
names.insert( 1, "Vasu" )
ages.insert( 1 , 18 )
print( names )
print( ages )
names.extend( ["Ishika" , "Arya"] )
print(names)
names.remove("Vani")
print(names)
remove = names.pop(-1)
print( remove )
print( names )
print( len(ages) )
print( ages.count(18) )
print( ages )
```

#Nested List

```python
Students = [
    ["Sheyali", 21, "AA" ] ,
    [ "Vishal" , 21, "AB"] ,
    ["Samridhhi" , 18 , "AD" ]
]
print( Students[2] )
Students[2][1] = 22
print( Students[2] )
Students.append([ 'Sachin' , 21 , 'AC' ])
print(Students)
```

```python
for Student in Students:
  print(Student)
```

```python
marks = [ 32, 40 , 35, 12, 16 ]
for mark in marks:
  if mark >= 15 :
    print( mark , "Pass" )
  else:
    print( mark , "Fail" )
```
